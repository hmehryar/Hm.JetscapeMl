```python
import tensorflow as tf

# Example categorical labels to indices (you need to define this mapping)
label_to_index = {'0.2': 0, '0.3': 1, '0.4': 2}  # Example mapping
num_classes = len(label_to_index)  # Number of unique classes

# Function to convert string label to one-hot vector
def encode_label(x, y):
    # Convert the string label to integer using the mapping
    y_index = tf.cast(label_to_index[y.numpy()], tf.int32)
    # One-hot encode the index
    y_one_hot = tf.one_hot(y_index, depth=num_classes)
    return x, y_one_hot

# Wrap the function for compatibility with tf.data (because of numpy dependency)
def encode_label_tf(x, y):
    return tf.py_function(func=encode_label, inp=[x, y], Tout=(tf.float64, tf.float32))
```

```python
dataset = example_dataset.map(encode_label_tf,
                            num_parallel_calls=tf.data.AUTOTUNE)
```

```python
def normalize_img(image, label):
  return (tf.cast(image, tf.float32) / max_value, label)

example_dataset = example_dataset.map(normalize_img,
                                      num_parallel_calls=tf.data.AUTOTUNE)

example_dataset
```
