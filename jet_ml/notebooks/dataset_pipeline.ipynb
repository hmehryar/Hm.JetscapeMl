```python
import tensorflow as tf

# Example categorical labels to indices (you need to define this mapping)
label_to_index = {'0.2': 0, '0.3': 1, '0.4': 2}  # Example mapping
num_classes = len(label_to_index)  # Number of unique classes

# Function to convert string label to one-hot vector
def encode_label(x, y):
    # Convert the string label to integer using the mapping
    y_index = tf.cast(label_to_index[y.numpy()], tf.int32)
    # One-hot encode the index
    y_one_hot = tf.one_hot(y_index, depth=num_classes)
    return x, y_one_hot

# Wrap the function for compatibility with tf.data (because of numpy dependency)
def encode_label_tf(x, y):
    return tf.py_function(func=encode_label, inp=[x, y], Tout=(tf.float64, tf.float32))
```

### storing the events into image directories

```python
import os
import tensorflow as tf
import jet_ml.dataset as ds
from PIL import Image  # For saving images as files

# Step 1: Load the dataset
dataset_size = 1000
(x, y) = ds.load_dataset(size=dataset_size)

print(f"Dataset shapes - x: {x.shape}, y: {y.shape}")

# Step 2: Combine y columns into a single string label
def combine_labels(y):
    # Join all 3 columns with underscores (e.g., "cat_small_fast")
    combined_labels = ["_".join(map(str, row)) for row in y]
    return combined_labels

# Combine labels
y_combined = combine_labels(y)

# Step 3: Define directories to save the images
output_root_dir = Config().DATA_DIR /"output_images"  # Root directory for output
display(output_root_dir)
unique_classes = list(set(y_combined))  # Extract unique labels from y
display(unique_classes)
# Create a directory for each class
for label in unique_classes:
    label_dir = os.path.join(output_root_dir, label)
    os.makedirs(label_dir, exist_ok=True)

# Step 3: Save each image into the respective directory
def save_images(x, y, root_dir):
    for i, (image, label) in enumerate(zip(x, y_combined)):
          # Ensure the label is a string
        if not isinstance(label, str):
            label = str(label)
        # Define the path for saving the image
        label_dir = os.path.join(root_dir, label)
        file_path = os.path.join(label_dir, f"image_{i:05d}.png")
        
        # Convert image data to PIL Image format and save
        img = Image.fromarray((image * 255).astype('uint8'))  # Rescale and convert to uint8
        img.save(file_path)
        
        # Print progress
        if i % 1000 == 0:
            print(f"Saved {i} images...")

# Step 4: Execute the saving function
save_images(x, y, output_root_dir)

print("All images have been saved into their respective directories.")
```

```python
def get_label(file_path):
    import os
    parts = tf.strings.split(file_path, os.path.sep)
    
    return parts[-2]
def get_separated_label(file_path):
    label=get_label(file_path)# Extract the combined label (e.g., 'MLBT_0.2_1.5')
    separated_label = tf.strings.split(label, "_")  # Split the label on "_"
    return separated_label
```

```python
get_separated_label(str(Config().DATA_DIR / "output_images/MLBT_0.2_1.5/image_00000.png"))
```

```python
def process_image(file_path):
    label = get_separated_label(file_path)
    img = tf.io.read_file(file_path) # load the raw data from the file as a string
    img = tf.image.decode_jpeg(img)
    # img = tf.image.resize(img, [128, 128])
    return img, label
```

```python
img, label = process_image(str(Config().DATA_DIR / "output_images/MLBT_0.2_1.5/image_00000.png"))
display(img.shape)
display(label)
```

```python
train_ds = train_ds.map(process_image)
test_ds = test_ds.map(process_image)
```

```python
for image, label in train_ds.take(1):
    print("****",image)
    print("****",label)
```

```python
import tensorflow as tf

def find_max_value(dataset):
    max_value = tf.constant(float('-inf'))  # Initialize to the smallest possible value
    for x, _ in dataset:  # Assuming the dataset yields (x, y)
        batch_max = tf.reduce_max(x)  # Find max in the current batch
        max_value = tf.maximum(max_value, batch_max)  # Update the global max
    return max_value

# Find max value for train_ds and test_ds
train_max = find_max_value(train_ds)
test_max = find_max_value(test_ds)

# Combine to get the overall max
overall_max = tf.maximum(train_max, test_max)

# Print results
print("Max value in train_ds:", train_max.numpy())
print("Max value in test_ds:", test_max.numpy())
print("Overall max value:", overall_max.numpy())
```

```python
def scale(image, label, max_value):
    return image/max_value, label
```

```python
train_ds = train_ds.map(lambda image, label: scale(image, label, overall_max))
```
