```python
# Loading/Preparing Environment for simulation
from jet_ml.config import Config
dataset_size=7200000 #10800000 #1000000


classification_parameter="building_alpha_s_dataset"
model_name="with_constant_q0"
server_name="wsu_grid_v100_cpu_24_mem_256gb"
simulation_name=f"{classification_parameter}_{model_name}_{int(dataset_size/1000)}k_dataset_size_{server_name}"

config=Config(simulation_name=simulation_name)
config.__str__()
```

```python
# Step 1: Load the dataset using the provided method
import jet_ml.dataset as ds 

(x, y)=ds.load_dataset(size=dataset_size)
```

```python
import numpy as np
# Step 2: Find unique vectors in combined_y and their indices
unique_vectors, inverse_indices, unique_counts = np.unique(y[:,2], axis=0, return_inverse=True,return_counts=True)
# just find the unique vectors with the same second element

display(f"unique_vectors: {unique_vectors}" )
print(f"inverse_indices:{inverse_indices}")
display(f"unique_counts of each q0:{unique_counts}")
```

```python
# save each dataset to a separate pickle file
from jet_ml.dataset import get_labels_str

label_str_dict=get_labels_str()
```

```python
#processing each unique vector and storing the dataset
for vector_id in range(unique_vectors.shape[0]):
    class_indices = np.where(inverse_indices == vector_id)[0]
    display(class_indices)
    dataset_with_unique_q0=(x[class_indices],y[class_indices])
    display(f"q0={vector_id} | dataset.x.shape={dataset_with_unique_q0[0].shape} | dataset.y.shape={dataset_with_unique_q0[1].shape}")
    # show the first 5 rows of each  dataset
    print("First 5 rows of each dataset:")
    print(f"{dataset_with_unique_q0[1][:5]}")
    # show the number of each unique alpha_s values in each dataset
    dataset_unique_vectors, dataset_inverse_indices,dataset_unique_counts =np.unique(dataset_with_unique_q0[1][:,1], axis=0, return_inverse=True,return_counts=True)
    # print len of each uniqdataset_inverse_indices

    display(f"q_0={vector_id}: alpha_s: {dataset_unique_vectors} counts: {dataset_unique_counts}", ) 

    #saving the dataset
    size=dataset_with_unique_q0[0].shape[0]
    dataset_file_name = f"jet_ml_benchmark_config_01_to_09_alpha_{label_str_dict['alpha_s_items_str']}_q0_{unique_vectors[vector_id]}_{label_str_dict['eloss_items_str']}_size_{size}_balanced_unshuffled.pkl"

    from jet_ml.dataset import save_dataset
    
    save_dataset(f"{config.DATA_DIR}/{dataset_file_name}",dataset_with_unique_q0)
    print(f"Dataset saved to {config.DATA_DIR}/{dataset_file_name}")
    
    
```
