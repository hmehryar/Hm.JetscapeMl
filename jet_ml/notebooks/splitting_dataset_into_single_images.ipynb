### storing the events into image directories

```python
import os
import tensorflow as tf
import jet_ml.dataset as ds
import numpy as np
from IPython.display import display

# Step 1: Load the dataset
(x, y) = ds.load_dataset(size=dataset_size)

print(f"Dataset shapes - x: {x.shape}, y: {y.shape}")

# Step 2: Combine y columns into a single string label
def combine_labels(y):
    # Join all 3 columns with underscores (e.g., "cat_small_fast")
    combined_labels = ["_".join(map(str, row)) for row in y]
    return combined_labels

# Combine labels
y_combined = combine_labels(y)

# Step 3: Define directories to save the images
output_root_dir = Config().DATA_DIR /f"jet_ml_benchmark_config_01_to_09_alpha_0.2_0.3_0.4_q0_1.5_2.0_2.5_MMAT_MLBT_size_{dataset_size}_balanced_unshuffled"  # Root directory for output
display(output_root_dir)
unique_classes = list(set(y_combined))  # Extract unique labels from y
display(unique_classes)
# Create a directory for each class
for label in unique_classes:
    label_dir = os.path.join(output_root_dir, label)
    os.makedirs(label_dir, exist_ok=True)

# Step 3: Save each image into the respective directory
def save_images(x, y, root_dir):
    for i, (image, label) in enumerate(zip(x, y_combined)):
          # Ensure the label is a string
        if not isinstance(label, str):
            label = str(label)
        # Define the path for saving the image
        label_dir = os.path.join(root_dir, label)
        os.makedirs(label_dir, exist_ok=True)  # Ensure the label directory exists
        file_path = os.path.join(label_dir, f"event_{i:07d}.npy")


        # Save the image data as a NumPy binary file
        np.save(file_path, image)
        
        # Print progress
        if i % 1000 == 0:
            print(f"Saved {i} event...")

# Step 4: Execute the saving function

# with tf.device("GPU"):
save_images(x, y, output_root_dir)

print("All images have been saved into their respective directories.")
```
